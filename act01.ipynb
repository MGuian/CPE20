## Part 3: Basic Signal Operations

```python
import numpy as np
import matplotlib.pyplot as plt


n_x = np.arange(0, 5)
x = np.array([2, -1, 3, 0, 4]) 

#1 Time Shift: y1[n] = x[n - 2]
n_y1 = n_x + 2
y1 = x.copy()

#2 Time Reversal: y2[n] = x[-n]
n_y2 = -n_x
y2 = x.copy()

#3. Energy: E = sum |x[n]|^2
E = np.sum(np.abs(x) ** 2)

print("Original Sequence: x[n] = {2, -1, 3, 0, 4}")
print("\tn_x =", n_x.tolist())
print("\tx =", x.tolist())
print("\n#1 Time Shift: y1[n] = x[n - 2]")
print("\tn_y1 =", n_y1.tolist())
print("\ty1 =", y1.tolist())
print("\n#2 Time Reversal: y2[n] = x[-n]")
print("\tn_y2 =", n_y2.tolist())
print("\ty2 =", y2.tolist())
print("\n#3. Energy: E = sum |x[n]|^2")
print("\tEnergy E =", E)
```

```python
#Original Sequence
plt.figure(figsize=(8, 4))
plt.stem(n_x, x)
plt.title('Original Sequence x[n]')
plt.xlabel('n (sample index)')
plt.ylabel('x[n]')
plt.grid(True)
plt.axhline(0, color='black', lw=0.5)
plt.axvline(0, color='black', lw=0.5)
plt.tight_layout()
plt.show()

#1 Time Shifted Sequence
plt.figure(figsize=(8, 4))
plt.stem(n_y1, y1)
plt.title('Time Shifted Sequence y1[n] = x[n - 2]')
plt.xlabel('n (sample index)')
plt.ylabel('y1[n]')
plt.grid(True)
plt.axhline(0, color='black', lw=0.5)
plt.axvline(0, color='black', lw=0.5)
plt.tight_layout()
plt.show()

#2 Time Reversed Sequence
plt.figure(figsize=(8, 4))
plt.stem(n_y2, y2)
plt.title('Time Reversed Sequence y2[n] = x[-n]')
plt.xlabel('n (sample index)')
plt.ylabel('y2[n]')
plt.grid(True)
plt.axhline(0, color='black', lw=0.5)
plt.axvline(0, color='black', lw=0.5)
plt.tight_layout()
plt.show()
```
